```python
import sqlite3
import pandas as pd
from pathlib import Path

conn = sqlite3.connect("../data/clean/trials.db")

out_dir = Path("../data/powerbi")
out_dir.mkdir(exist_ok=True)

exports = {
    "01_studies_per_year.sql": "studies_per_year.csv",
    "02_competition_by_area_phase.sql": "competition_area_phase.csv",
    "03_stop_rate_by_phase.sql": "stop_rate_by_phase.csv",
    "04_city_concentration.sql": "city_concentration.csv",
    "04b_studies_per_city.sql": "studies_per_city.csv",
    "05_top_industry_sponsors.sql": "top_industry_sponsors.csv",
    "06_sponsor_class_split.sql": "sponsor_class_split.csv",
    "07_top_named_centres.sql": "top_named_centres.csv"
}

for sql_file, csv_file in exports.items():
    query = (Path("../sql") / sql_file).read_text(encoding="utf-8")
    df = pd.read_sql(query, conn)
    df.to_csv(out_dir / csv_file, index=False)
    print(csv_file, df.shape)

conn.close()
```
