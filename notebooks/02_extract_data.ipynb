```python
import requests
import pandas as pd
import json

url = "https://clinicaltrials.gov/api/v2/studies"
```

```python
params = {"query.locn": "Romania", "pageSize": 1000, "countTotal": "true"}
all_studies = []

while True:
    r = requests.get(url, params=params)
    data = r.json()
    all_studies.extend(data["studies"])
    token = data.get("nextPageToken")
    if token is None:
        break
    params["pageToken"] = token

print(len(all_studies))
```

```python
with open("raw_studies.json", "w") as f:
    json.dump(all_studies, f)
```
